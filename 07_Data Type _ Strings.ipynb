#### Slice stride

```python
numbers = '0123456789'
print(f'All numbers: {numbers[::]}')

print(f'Every even number: {numbers[::2]}')
print(f'Every odd number: {numbers[1::2]}')

print(f'Every third number between 1 and 8: {numbers [1:9:3]}')
```
